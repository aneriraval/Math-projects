```python
#Ex 4.2:Q3
import numpy as np
import matplotlib.pyplot as plt
import sympy as py
x=np.array([46,49,51,52,54,56,57,58,59,60,61,62,63,64,66,67,68,71,72])
y=np.array([40,50,55,63,72,70,77,73,90,93,96,88,99,110,113,120,127,137,132])
yp=0
xplt=np.linspace(x[0],x[-1],300) #creating 300 evenly spaced x-values from 46 to 72
yplt=[]
#Lagrangian Interpolation:
for xp in xplt:
    yp = 0
    for xi, yi in zip(x, y):
        li = np.prod((xp - x[x != xi]) / (xi - x[x != xi]))
        yp += yi * li
    yplt.append(yp)

yplt = np.array(yplt)

plt.plot(xplt, yplt, label='Lagrange Interpolation')
plt.plot(x, y, 'ro', label='Data Points')
plt.legend()
plt.grid(True)
plt.title("Lagrange Interpolation Polynomial")
plt.xlabel("Temperature (in F)")
plt.ylabel("No.of times a cricket chirps in 1 minute")
plt.ylim(-2.8*10**6.5,1500000) #adjusting range
plt.show()
```

```python
#Ex 4.3:Q8

import numpy as np
import matplotlib.pyplot as plt

# Data
year = np.array([
    1790, 1800, 1810, 1820, 1830, 1840, 1850, 1860, 1870, 1880,
    1890, 1900, 1910, 1920, 1930, 1940, 1950, 1960, 1970, 1980,
    1990, 2000])

population = np.array([
    3929000, 5308000, 7240000, 9638000, 12866000, 17069000, 23192000, 31443000, 38558000, 50156000,
    62948000, 75995000, 91972000, 105711000, 122755000, 131669000, 150697000, 179323000,
    203212000, 226505000, 248709873, 281416000])

x = year - 1790
y = population
coeffs = np.polyfit(x, y, 2)
a, b, c = coeffs
print(f"Quadratic model: P(x) = {a:.5f}x^2 + {b:.5f}x + {c:.5f}")
y_fit = a * x**2 + b * x + c
x_plot = np.linspace(0, 210, 300)
y_plot = a * x_plot**2 + b * x_plot + c
# Plot
plt.scatter(x, y, color='red', label='Data')
plt.plot(x_plot, y_plot, label='Quadratic Fit')
plt.xlabel("Years since 1790")
plt.ylabel("Population")
plt.title("US Population 1790-2000 Quadratic Fit")
plt.legend()
plt.grid(True)
plt.show()

y_mean = np.mean(y)
ssr = np.sum((y_fit - y_mean)**2)
sst = np.sum((y - y_mean)**2)
r_squared = ssr / sst
print("R² =", r_squared)
```
