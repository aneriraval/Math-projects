```python
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)
X = np.linspace(0, 10, 50) #Generating 50 evenly spaced points between 1 and 10
y = 2 * X + 5 + np.random.normal(0, 2, len(X))

# Gradient Descent Parameters
m = 0.0          # Initial guess for slope
c = 0.0          # Initial guess for intercept
alpha = 0.01     # Learning rate (step size)
iterations = 2000 # Number of steps to reach the minima
n = float(len(X))

#The Gradient Descent Method
for i in range(iterations):
    y_pred = m * X + c

    # Calculate Gradients (Least squares)
    dm = (-2/n) * np.sum(X * (y - y_pred))
    dc = (-2/n) * np.sum(y - y_pred)
    # Update parameters to move toward the local minima
    m = m - alpha * dm
    c = c - alpha * dc

# Results
print(f"Local Minima found using Gradient Descent:")
print(f"Final Slope (m): {m:.4f}")
print(f"Final Intercept (c): {c:.4f}")

# Plotting the results
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', label='Data Points')
plt.plot(X, m * X + c, color='red', linewidth=2,
         label=f'GD Minima Fit: y = {m:.2f}x + {c:.2f}')

plt.title("Least Squares: Finding Local Minima via Gradient Descent")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

#Number generation
np.random.seed(42)
X = np.linspace(0, 10, 50)
y = 2 * X + 5 + np.random.normal(0, 2, len(X))
m, c = 0.0, 0.0
alpha = 0.01
iterations = 2000
n = float(len(X))
loss_history = []
for i in range(iterations):
    y_pred = m * X + c
    mse = np.mean((y - y_pred)**2)
    loss_history.append(mse)
    dm = (-2/n) * np.sum(X * (y - y_pred))
    dc = (-2/n) * np.sum(y - y_pred)
    m -= alpha * dm
    c -= alpha * dc
plt.figure(figsize=(10, 5))
plt.plot(range(iterations), loss_history, color='purple', lw=2)
plt.title("Loss Convergence (MSE over Iterations)")
plt.xlabel("Iteration Number")
plt.ylabel("Mean Squared Error")
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()

print(f"Final Loss: {loss_history[-1]:.4f}")
```
